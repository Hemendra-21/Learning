### **📜 What are Data Contracts?**

A **data contract** is a **formal agreement** between a **data producer** (who creates the data) and **data consumers** (who use the data).

It clearly defines **what data is provided, how it looks, how it behaves, and what guarantees are made.**

👉 Think of it like an **API contract**, but for **data** instead of services.

---------------

**🧠 Simple Definition**

> A data contract specifies **schema, semantics, quality rules, freshness, and ownership** of a dataset so producers and consumers can evolve independently **without breaking each other**.


-------------------

**🔄 Why Data Contracts Exist (The Core Problem)**

Without data contracts:

- Columns get renamed or deleted ❌

- Data types change silently ❌

- Nulls appear unexpectedly ❌

- Downstream dashboards & ML models break ❌

- No one knows who owns the data ❌

**With data contracts:**

- Changes are intentional and controlled ✅

- Expectations are explicit ✅

- Failures are caught early ✅

- Trust in data increases ✅

---------------

#### **🧩 What a Data Contract Typically Defines**

**1️⃣ Schema**

- Column names

- Data types

- Nullable vs non-nullable

- Primary keys

```python
columns:
  - name: customer_id
    type: integer
    nullable: false
  - name: email
    type: string
    nullable: false
```

--------

**2️⃣ Semantics (Meaning of Data)**

Explains **what the data represents**, not just its type.

```python
description:
  customer_id: "Unique identifier for a customer"
  email: "Customer's primary email address used for login"
```

---------------

**3️⃣ Data Quality Rules**

Rules that must always hold true.

Examples:

- `customer_id` must be unique

- `email` must not be null

- `status` must be one of specific values

```python
quality:
  - rule: not_null
    column: customer_id
  - rule: unique
    column: customer_id
  - rule: accepted_values
    column: status
    values: ["active", "inactive"]
```

----------------

**4️⃣ Freshness & SLAs**

Defines **how fresh** the data should be.

```python
freshness:
  max_delay: 24h
```

👉 If data is older than 24 hours → contract violation 🚨

----------------

**5️⃣ Ownership & Accountability**

Who is responsible when things go wrong?

```python
owner:
  team: analytics-platform
  contact: analytics@company.com
```

-----------

**6️⃣ Versioning & Change Policy**

How changes are handled.

- Backward-compatible changes → allowed

- Breaking changes → new version

```python
version: v1
change_policy:
  breaking_changes: require_new_version
```

**🏗️ Data Contract Flow (End-to-End)**

1) **Producer defines contract**

2) Consumer agrees to contract

3) Data is produced

4) Contract is validated (schema + quality)

5) If violated → pipeline fails / alert sent

6) Consumers trust the data

--------------


**🧪 Data Contracts vs dbt Tests (Very Important)**

| Concept       | Data Contracts             | dbt Tests          |
| ------------- | -------------------------- | ------------------ |
| Purpose       | Agreement                  | Validation         |
| Scope         | Schema + meaning + SLA     | Data correctness   |
| Who defines   | Producer                   | Analytics engineer |
| When enforced | Before/at consumption      | During dbt runs    |
| Example       | "email must never be null" | `not_null(email)`  |


**👉 dbt tests are one way to enforce a data contract**, but a contract is **bigger than tests.**

--------------


#### **🧠 Data Contracts in dbt (Practical Mapping)**

In dbt, data contracts are usually expressed using:

- `schema.yml`

- Model contracts (`contracts: enforced: true`)

- Tests

- Documentation

**Example (dbt-style contract)**

```python
models:
  - name: dim_customers
    contracts:
      enforced: true
    columns:
      - name: customer_id
        data_type: integer
        tests:
          - not_null
          - unique
      - name: email
        data_type: string
        tests:
          - not_null
```

👉 If schema or data breaks → **dbt fails immediately**

---------------

**🚨 What Happens When a Contract Is Broken?**

- Pipeline fails

- Alerts sent

- Data not promoted to downstream layers

- Consumers are protected

This is called **fail fast, fail safe.**

-----------------------------


**🧠 Best Practices for Data Contracts**

✅ Define contracts **at the source**

✅ Keep them **human-readable (YAML)**

✅ Enforce automatically (CI/CD, dbt)

✅ Version contracts

✅ Treat breaking changes seriously

✅ Make ownership explicit
