**Documentation in dbt** is how you **describe, organize, and visualize your data models** so that anyone (analysts, engineers, stakeholders) can understand:

- what each model represents

- where the data comes from

- how models are related (lineage)

- what each column means

- how trustworthy the data is (via tests)

dbt turns this written metadata into a **live documentation website**.

---------------

#### **1️⃣ Why dbt documentation matters**

Think of dbt docs as **self-updating data documentation.**

Without docs:

- Tables are confusing

- Column meanings are tribal knowledge

- Debugging is slow

With docs:

- Every model is explained

- Column-level clarity

- Lineage shows upstream → downstream impact

- Tests appear alongside documentation

👉 This is one of the **biggest strengths of dbt.**

------------


#### **2️⃣ Where documentation is written in dbt**

Documentation is written mainly in **YAML files** (usually `schema.yml` or `<model_name>.yml`).

Typical locations:

```python
models/
 ├── staging/
 │    └── schema.yml
 ├── marts/
 │    └── schema.yml
```

----------

#### **3️⃣ Model-level documentation**

You document **what a model** is using description.

**Example**

```python
version: 2

models:
  - name: dim_customers
    description: "Dimension table containing one row per customer."
```

This description appears in:

- dbt docs website

- lineage graph

- compiled metadata

-----------

#### **4️⃣ Column-level documentation (very important)**

This explains **what each column means.**

**Example**

```python
models:
  - name: dim_customers
    description: "Customer dimension table."
    columns:
      - name: customer_id
        description: "Unique identifier for each customer."

      - name: email
        description: "Customer email address."

      - name: created_at
        description: "Timestamp when the customer record was created."
```

✅ Best practice: **document every column in marts & dimensions**

-------------

#### **5️⃣ Documentation + Tests (together)**

In dbt, **documentation and tests live side by side.**

```python
models:
  - name: dim_customers
    description: "Customer dimension table."
    columns:
      - name: customer_id
        description: "Primary key for customers."
        tests:
          - unique
          - not_null
```

📌 Result in docs:

- Column description shown

- Tests displayed → builds trust in data

----------

#### **6️⃣ Source documentation**

You can document **raw/source** tables too.

```python
sources:
  - name: airbnb
    description: "Raw data from Airbnb system"
    tables:
      - name: hosts
        description: "Raw hosts data ingested from Airbnb"
```

This helps answer:

> “Where did this data come from?”

----------------

#### **7️⃣ Generating dbt documentation**

Once documentation is written, generate it using:

```python
dbt docs generate

dbt docs serve
```

What happens:

- dbt reads models + YAML files

- Builds metadata (descriptions, tests, lineage)

- Launches a **web UI**

--------------

#### **8️⃣ What you see in dbt docs UI**

The documentation site includes:

**🔹 Models page**

- Model descriptions

- Columns + data types

- Tests


**🔹 Lineage graph (DAG)**

- Upstream dependencies (source, ref)

- Downstream usage

- Impact analysis

**🔹 Sources page**

- Raw table documentation

- Freshness info (if configured)

--------------

#### **9️⃣ Documentation using `doc()` blocks (advanced but powerful)**

Reusable documentation using Markdown.

**Step 1: Define docs**

Create `docs.md`

```python
{% docs customer_id %}
A unique identifier assigned to each customer.
{% enddocs %}
```

**Step 2: Use it in YAML**

```python
columns:
  - name: customer_id
    description: "{{ doc('customer_id') }}"
```

✅ Reusable

✅ Centralized

✅ Clean YAML files

----------

#### **🔟 Best practices for dbt documentation ⭐**

- Document **all models in marts**

- Always document **primary keys**

- Keep descriptions **business-friendly**

- Use **doc()** blocks for repeated definitions

- Keep tests close to documentation

- Treat docs as **production assets**, not optional
