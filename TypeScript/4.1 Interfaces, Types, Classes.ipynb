**1. Differences: interface vs. type**

*Example 1: Defining Object Shapes*

- `interface` is used for defining object shapes and contracts.

```python
interface User {
  name: string;
  age: number;
}

const user: User = {
  name: "Alice",
  age: 25,
};
```

- `type` can also define object shapes but is more flexible.

```python
type User = {
  name: string;
  age: number;
};

const user: User = {
  name: "Bob",
  age: 30,
};
```

*Example 2: Union Types (Only with `type`)*

`type` can define unions, which `interface` cannot.

```python
type Status = "success" | "error" | "loading";

function handleStatus(status: Status) {
  if (status === "success") {
    console.log("Operation succeeded!");
  }
}

handleStatus("success"); // ✅
handleStatus("error");   // ✅
```

*Example 3: Declaration Merging (Only with interface)*

`interface` supports declaration merging, allowing you to add properties to an existing interface.



```python
interface Product {
  name: string;
}

interface Product {
  price: number;
}

const product: Product = {
  name: "Laptop",
  price: 1000,
};
```

`type` does not support declaration merging.

**2. Use Cases and Best Practices**

*Use Case 1: API Response Models*

Use `interface` for defining API response models because they are extendable and can be merged.



```python
interface ApiResponse {
  status: string;
  message: string;
}

interface UserResponse extends ApiResponse {
  data: {
    id: number;
    name: string;
  };
}

const response: UserResponse = {
  status: "success",
  message: "User fetched successfully",
  data: {
    id: 1,
    name: "Alice",
  },
};
```

*Use Case 2: Complex Type Combinations*

Use `type` for unions or intersections when combining multiple types.



```python
type ErrorResponse = {
  status: "error";
  error: string;
};

type SuccessResponse = {
  status: "success";
  data: any;
};

type ApiResponse = ErrorResponse | SuccessResponse;

function handleApiResponse(response: ApiResponse) {
  if (response.status === "error") {
    console.error(response.error);
  } else {
    console.log(response.data);
  }
}
```

*3. Extending Interfaces*

Example: Extending Interfaces

You can extend an `interface` using the `extends` keyword.

```python
interface Vehicle {
  make: string;
  model: string;
}

interface Car extends Vehicle {
  doors: number;
}

const myCar: Car = {
  make: "Toyota",
  model: "Corolla",
  doors: 4,
};
```

*Use Case: Extending for Reusability*

Use `extends` to create reusable interfaces for shared properties.



```python
interface BaseEntity {
  id: number;
  createdAt: Date;
}

interface User extends BaseEntity {
  name: string;
  email: string;
}

interface Product extends BaseEntity {
  name: string;
  price: number;
}

const user: User = {
  id: 1,
  createdAt: new Date(),
  name: "Alice",
  email: "alice@example.com",
};

const product: Product = {
  id: 101,
  createdAt: new Date(),
  name: "Laptop",
  price: 1200,
};
```

**4. Combining Types with Intersections**

*Example: Combining Types*

Use the `&` operator to combine multiple types into one.

```python
type Person = {
  name: string;
  age: number;
};

type Employee = {
  employeeId: number;
  department: string;
};

type EmployeeDetails = Person & Employee;

const employee: EmployeeDetails = {
  name: "John",
  age: 35,
  employeeId: 1234,
  department: "Engineering",
};
```

**Use Case: Combining Features**

Combine multiple types to represent objects with shared and unique properties.



```python
type Timestamps = {
  createdAt: Date;
  updatedAt: Date;
};

type BlogPost = {
  title: string;
  content: string;
};

type BlogPostWithTimestamps = BlogPost & Timestamps;

const post: BlogPostWithTimestamps = {
  title: "Understanding TypeScript",
  content: "TypeScript is a superset of JavaScript...",
  createdAt: new Date(),
  updatedAt: new Date(),
};
```

**5. Advanced Example: Combining Interfaces and Types**

You can mix `interface` and `type` using intersections.

```python
interface Address {
  street: string;
  city: string;
}

type ContactInfo = {
  phone: string;
  email: string;
};

type UserProfile = Address & ContactInfo & {
  name: string;
  age: number;
};

const profile: UserProfile = {
  name: "Alice",
  age: 28,
  street: "123 Main St",
  city: "Wonderland",
  phone: "123-456-7890",
  email: "alice@example.com",
};
```

**Best Practices**

Use `interface` for object shapes and extending contracts.

Use `type` for unions, intersections, or complex type combinations.

Prefer `interface` when declaration merging is required.

Use `type` for flexibility when combining multiple types.
