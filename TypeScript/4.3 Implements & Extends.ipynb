In TypeScript, `implements` and `extends` are used to define relationships between classes and interfaces or other classes.



**1. Implementing Interfaces in Classes**

When a class `implements` an interface, it ensures that the class adheres to the structure defined by the interface. This is useful for enforcing a contract that the class must follow.

*Example:*

```python
// Define an interface
interface Animal {
    name: string;
    makeSound(): void;
}

// Implement the interface in a class
class Dog implements Animal {
    name: string;

    constructor(name: string) {
        this.name = name;
    }

    makeSound(): void {
        console.log(`${this.name} says: Woof!`);
    }
}

const myDog = new Dog("Buddy");
myDog.makeSound(); // Output: Buddy says: Woof!
```

- The `Dog` class must implement all properties and methods defined in the `Animal` interface.

- If any property or method is missing, TypeScript will throw an error.

-------------


**2. Extending Classes**

When a class `extends` another class, it inherits all properties and methods from the parent class. The child class can also override or add new functionality.

*Example:*

```python
// Define a base class
class Animal {
    name: string;

    constructor(name: string) {
        this.name = name;
    }

    makeSound(): void {
        console.log(`${this.name} makes a sound.`);
    }
}

// Extend the base class
class Dog extends Animal {
    constructor(name: string) {
        super(name); // Call the parent class constructor
    }

    // Override the makeSound method
    makeSound(): void {
        console.log(`${this.name} says: Woof!`);
    }
}

const myDog = new Dog("Buddy");
myDog.makeSound(); // Output: Buddy says: Woof!
```

- The `Dog` class inherits the `name` property and `makeSound` method from the Animal class.

- The `super` keyword is used to call the parent class constructor.
--------


**3. Extending Interfaces**

Interfaces can also extend other interfaces. This allows you to build on existing structures.

*Example:*


```python
// Define a base interface
interface Animal {
    name: string;
    makeSound(): void;
}

// Extend the base interface
interface Pet extends Animal {
    owner: string;
}

// Implement the extended interface
class Cat implements Pet {
    name: string;
    owner: string;

    constructor(name: string, owner: string) {
        this.name = name;
        this.owner = owner;
    }

    makeSound(): void {
        console.log(`${this.name} says: Meow!`);
    }
}

const myCat = new Cat("Whiskers", "Alice");
myCat.makeSound(); // Output: Whiskers says: Meow!
console.log(`Owner: ${myCat.owner}`); // Output: Owner: Alice
```

- The `Pet` interface extends `Animal`, adding an `owner` property.

- The `Cat` class implements the `Pet` interface, adhering to both `Animal` and `Pet` requirements.

---------


**4. Combining Implements and Extends**

You can combine `implements` and `extends` in a single class.

*Example:*

```python
interface Animal {
    name: string;
    makeSound(): void;
}

class Mammal {
    hasFur: boolean;

    constructor(hasFur: boolean) {
        this.hasFur = hasFur;
    }
}

// A class that extends a base class and implements an interface
class Dog extends Mammal implements Animal {
    name: string;

    constructor(name: string, hasFur: boolean) {
        super(hasFur); // Call the parent class constructor
        this.name = name;
    }

    makeSound(): void {
        console.log(`${this.name} says: Woof!`);
    }
}

const myDog = new Dog("Buddy", true);
myDog.makeSound(); // Output: Buddy says: Woof!
console.log(`Has fur: ${myDog.hasFur}`); // Output: Has fur: true
```

- The `Dog` class extends the `Mammal` class and implements the `Animal` interface.

- This allows the class to inherit functionality from the parent class while adhering to the interface contract.
