**Type Assertions**

Type assertions in TypeScript allow you to override the inferred type of a value when you, as the developer, know more about the type than TypeScript does.

They are useful in scenarios where TypeScript cannot infer the type correctly or when working with third-party libraries.

Syntax

1) `value as string` (Preffered in Modern TypeScript)



```python
let someValue: unknown = "Hello, TypeScript!";
let strLength: number = (someValue as string).length;
```

2) `<string>value` (Alternative syntax, not recommended in JSX files)

```python
let someValue: unknown = "Hello, TypeScript!";
let strLength: number = (<string>someValue).length;
```

----------

**When to Use Type Assertions**

- When TypeScript cannot infer the type:



```python
const canvas = document.getElementById("myCanvas") as HTMLCanvasElement;
const ctx = canvas.getContext("2d");
```

- When working with third-party libraries:

```python
const response = fetchData() as MyResponseType;
```

- When narrowing `unknown` or `any` types:

```python
function processValue(value: unknown) {
    if (typeof value === "string") {
        console.log((value as string).toUpperCase());
    }
}
```

When NOT to Use Type Assertions

- When TypeScript already knows the type:



```python
let name: string = "John";
// Unnecessary: let nameLength = (name as string).length;
let nameLength = name.length; // Correct
```

- When it leads to unsafe assumptions:

```python
let input = document.getElementById("myInput") as HTMLInputElement;
input.value = "Hello"; // Unsafe if `myInput` is not an input element
```

- Avoid using assertions to bypass errors:

```python
let value: any = 123;
let str = value as string; // Unsafe, may cause runtime errors
```

Best Practices
- Use type assertions sparingly and only when you are confident about the type.

- Prefer `value as Type` syntax for consistency and compatibility with JSX.
- Avoid using assertions to suppress TypeScript errors without understanding the root cause.
