2.2 Object Types

1) `object`

The object type represents any non-primitive value (e.g., arrays, functions, or objects). However, it is rarely used directly because it doesn't provide much type safety.

Example:



```python
let obj: object = { name: "Alice", age: 25 };

// Use Case:
// Use object when you want to allow any non-primitive value but don't need to specify its structure.

```

2) `Custom Object Types`

You can define the structure of an object using interfaces or type aliases.

Example:

Interface :

- Interfaces are specifically designed to define object types.

- They support features like declaration merging, which allows you to extend or add properties to an existing interface.

Example:


```python
interface Employee{
    name : string;
    salary: number,
    isActive : boolean;

}


let user:User = {
    name:"Hemendra",
    salary:50000,
    isActive: true,
};

// Use Case:
// Use custom object types to define the shape of data for better type safety and code clarity.
```

Type Aliases :


```python
type Employee = {
    name: string;
    salary: number;
    isActive: boolean;
};

let user: Employee = {
    name: "Hemendra",
    salary: 50000,
    isActive: true,
};
```

Conclusion:

Type aliases can define object types, just like interfaces. However, interfaces are more suited for defining object shapes due to their additional features like declaration merging. Use type aliases when you need more flexibility (e.g., union or intersection types).

Optional and readonly properties

```python
// Optional Properties

// Optional properties are properties that may or may not be present in an object. They are denoted using a ? after the property name.

// Example:

interface Employee {
    name: string;
    salary: number;
    isActive?: boolean; // Optional property
}

let employee1: Employee = {
    name: "Alice",
    salary: 50000,
};

let employee2: Employee = {
    name: "Bob",
    salary: 60000,
    isActive: true, // Optional property is included
};

// In the above example, the isActive property is optional. It can be omitted when creating an object of type Employee.
```

```python
// Readonly Properties
// Readonly properties are properties that cannot be modified after the object is created. They are denoted using the readonly keyword.

// Example:

interface Employee {
    readonly id: number; // Readonly property
    name: string;
    salary: number;
}

let employee: Employee = {
    id: 101,
    name: "Alice",
    salary: 50000,
};

// employee.id = 102; // Error: Cannot assign to 'id' because it is a read-only property
```

Combining Optional and Readonly

You can combine readonly and ? to create properties that are both optional and immutable.

Example:

```python
interface Employee {
    readonly id?: number; // Optional and readonly
    name: string;
    salary: number;
}

let employee: Employee = {
    name: "Alice",
    salary: 50000,
};

// employee.id = 102; // Error: Cannot assign to 'id' because it is a read-only property
```
