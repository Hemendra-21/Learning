### **✅ PHASE 8: Best Practices for Maintainable Code**

- Prefer `type` for complex unions, `interface` for public contracts

- Use `unknown` over `any` when unsure

- Avoid unnecessary type assertions

- Always enable `"strict": true` in tsconfig

- Avoid global type definitions

- Group types into folders: `types/`, `interfaces/`, `models/`

- Use linter + formatter (ESLint + Prettier with TS rules)

## 🧠 TypeScript Roadmap for Angular

| 🧵 Topic(s) | 📌 Why It’s Important for Angular |
|-------------|-----------------------------------|
| `types`, `interfaces` | Define object shapes, service response models, and component inputs for better type safety. |
| `classes`, `constructors`, `access modifiers (public/private/protected)` | Angular is class-based. Components, services, and DI (dependency injection) use these concepts. |
| `decorators` | Core Angular concept (`@Component`, `@Injectable`, etc.) — decorators come from TypeScript. |
| `generics` | Used in RxJS Observables and type-safe reusable services or functions. |
| `enums`, `type aliases`, `readonly`, `optional (?)`, `union types` | Essential for handling configuration, form validation, and defining robust type structures. |

✅ **Tip:** Build mini examples (like a todo model or mock API response) using each concept to reinforce your understanding.
