##### TypeScript allows you to define optional parameters and default parameters in functions to make them more flexible and user-friendly.

**1. Optional Parameters**

Optional parameters are denoted by a ? after the parameter name. These parameters are not required when calling the function.

Example:

```python
function greet(name?: string): string {
    return name ? `Hello, ${name}!` : "Hello!";
}

console.log(greet("Hemendra")); // Output: Hello, Hemendra!
console.log(greet());           // Output: Hello!
```

Key Points:

- Optional parameters are automatically assigned `undefined` if not provided.
- Use optional parameters when the argument is not always required.

-------------

**2. Default Parameters**

Default parameters allow you to assign a default value to a parameter if no argument is provided.

Example:

```python
function greet(name: string = "Guest"): string {
    return `Hello, ${name}!`;
}

console.log(greet("Hemendra")); // Output: Hello, Hemendra!
console.log(greet());           // Output: Hello, Guest!
```

Key Points:

- Default parameters are used when you want a fallback value.
- They are evaluated only if the argument is `undefined`.

--------

**3. Combining Optional and Default Parameters**

You can combine optional and default parameters, but default parameters should come after optional ones.

Example:

```python
function greet(name?: string, greeting: string = "Hello"): string {
    return name ? `${greeting}, ${name}!` : `${greeting}!`;
}

console.log(greet("Hemendra"));       // Output: Hello, Hemendra!
console.log(greet(undefined, "Hi"));  // Output: Hi!
console.log(greet());                 // Output: Hello!
```

Best Practices
- Use optional parameters for truly optional arguments.
- Use default parameters for fallback values.
- Avoid mixing optional and default parameters in a way that causes ambiguity.
