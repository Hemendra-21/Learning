**Module Resolution in TypeScript**

Module resolution is the process by which TypeScript determines the location of the files that are imported in your code. It is a critical part of how TypeScript works with modules.

Key Concepts
- 1) Module Import Syntax
    - TypeScript uses `import` statements to bring in modules.
    - Example:


```python
import { MyClass } from './myModule';
```

- 2) **Module Resolution Strategies** TypeScript provides two module resolution strategies:

    - **Classic**: Used for backward compatibility with older JavaScript projects.
    - **Node**: Mimics the Node.js module resolution mechanism.

    The strategy can be set in the `tsconfig.json` file:

```python
// {
//   "compilerOptions": {
//     "moduleResolution": "node" // or "classic"
//   }
// }
```

- 3) **Path Mapping**

    - You can define custom paths for module imports using the `paths` and `baseUrl` options in `tsconfig.json`.
    
    - *Example:*


```python
// {
//   "compilerOptions": {
//     "baseUrl": "./",
//     "paths": {
//       "@utils/*": ["src/utils/*"]
//     }
//   }
// }
```

- 4) **TypeScript's Resolution Process**

    - TypeScript first checks for a file with the exact name.
    - If not found, it looks for an index file in the folder (e.g., index.ts or index.js).
    - It also respects extensions like `.ts`, `.tsx`, `.d.ts`, and `.js`.
