##### 1. `any` vs. `unknown` (Why `unknown` is safer)

- `any`: Disables type checking, allowing any value to be assigned or accessed without restrictions. This can lead to runtime errors.

- `unknown`: A safer alternative to `any`. You must narrow the type before using it, ensuring type safety.

Example:



```python
let value: any;
value = 42;
value.toUpperCase(); // No error, but runtime crash if value is not a string.

let saferValue: unknown;
saferValue = 42;
// saferValue.toUpperCase(); // Error: Object is of type 'unknown'.

if (typeof saferValue === "string") {
    console.log(saferValue.toUpperCase()); // Safe to use after type check.
}
```

##### 2. `void` and `never`

- `void`: Represents the absence of a return value. Commonly used for functions that do not return anything.

- `never`: Represents a value that never occurs. Used for functions that throw errors or have infinite loops.

Example:

```python
// `void` example
function logMessage(message: string): void {
    console.log(message);
}

// `never` example
function throwError(message: string): never {
    throw new Error(message);
}

function infiniteLoop(): never {
    while (true) {}
}
```

##### 3. `enum` (Numeric and String Enums)

- **Numeric Enums**: Default values are auto-incremented numbers starting from 0.

- **String Enums**: Explicitly assign string values to enum members.

Example:

```python
// Numeric Enum
enum Direction {
    Up,    // 0
    Down,  // 1
    Left,  // 2
    Right  // 3
}
console.log(Direction.Up); // Output: 0

// String Enum
enum Status {
    Success = "SUCCESS",
    Failure = "FAILURE"
}
console.log(Status.Success); // Output: "SUCCESS"
```

If you don't assign any values to the properties of an `enum` in TypeScript, it will default to being a **numeric enum**.
The values will start from `0` and increment by `1` for each subsequent member.

Example:

```python
enum Direction {
    Up,    // 0
    Down,  // 1
    Left,  // 2
    Right  // 3
}

console.log(Direction.Up);    // Output: 0
console.log(Direction.Down);  // Output: 1
console.log(Direction.Left);  // Output: 2
console.log(Direction.Right); // Output: 3



// If you explicitly assign a value to one member, the subsequent members will increment from that value unless explicitly assigned otherwise.

// Example with Explicit Assignment:


enum Direction {
    Up = 5,    // 5
    Down,      // 6
    Left,      // 7
    Right      // 8
}

console.log(Direction.Up);    // Output: 5
console.log(Direction.Down);  // Output: 6
console.log(Direction.Left);  // Output: 7
console.log(Direction.Right); // Output: 8
```

##### Literal Types in TS

Literal types in TypeScript allow you to specify exact values a variable can hold. These can be `string literals`, `number literals`, or `boolean literals`. 

They are useful for creating more precise and constrained types.

```python
// String Literal Types

// You can restrict a variable to specific string values.

type Size = 'small' | 'medium' | 'large';

let shirtSize: Size;

shirtSize = 'medium'; // ✅ Valid
shirtSize = 'extra-large'; // ❌ Error: Type '"extra-large"' is not assignable to type 'Size'.
```

```python
// Number Literal Types
// You can restrict a variable to specific numeric values.

type DiceRoll = 1 | 2 | 3 | 4 | 5 | 6;

let roll: DiceRoll;

roll = 4; // ✅ Valid
roll = 7; // ❌ Error: Type '7' is not assignable to type 'DiceRoll'.
```

```python
// Boolean Literal Typestype 
// You can restrict a variable to true or false.

type IsActive = true;

let active: IsActive;

active = true; // ✅ Valid
active = false; // ❌ Error: Type 'false' is not assignable to type 'true'.
// You can restrict a variable to true or false.

```

----

**Combining Literal Types with Union Types**

Literal types are often combined with union types to define a set of allowed values.



```python
type Status = 'success' | 'failure' | 'pending';

function updateStatus(status: Status): void {
    console.log(`Status updated to: ${status}`);
}

updateStatus('success'); // ✅ Valid
updateStatus('error');   // ❌ Error: Type '"error"' is not assignable to type 'Status'.
```

**Literal Types with Enums**

Enums can also be used to define literal-like values.



```python
enum Direction {
    Up = 'UP',
    Down = 'DOWN',
    Left = 'LEFT',
    Right = 'RIGHT',
}

let move: Direction;

move = Direction.Up; // ✅ Valid
move = 'UP';         // ❌ Error: Type '"UP"' is not assignable to type 'Direction'.
```

Use Cases
- **Function Parameters**: Restrict input values to specific options.
- **State Management**: Define allowed states in an application.
- **Configuration**: Use for settings or options with predefined values.

Literal types improve type safety by ensuring only valid values are used in your code.
