### JavaScript Classes
- JavaScript classes are a blueprint for creating objects. 

- They were introduced in ES6 (ECMAScript 2015) and provide a cleaner, more intuitive syntax for working with prototypes and object-oriented programming (OOP) in JavaScript.


**Key Features of JavaScript Classes:**
- **Class Declaration**: Classes are defined using the `class` keyword.

- **Constructor Method**: A special method called `constructor` is used to initialize object properties.
- **Methods**: Functions inside a class are called methods. They can be instance methods or static methods.
- **Inheritance**: Classes can extend other classes using the `extends` keyword, enabling inheritance.
- **Encapsulation**: Classes support encapsulation using `public`, `private`, and `protected` (in TypeScript or with modern JS private fields).


---------

```python
// 1.Class Declaration

class Person {
    constructor(name, age) {
        (this.name = name), (this.age = age);
    }

    greet() {
        console.log(`Hello ${this.name}, You are of age ${this.age}`);
    }
}

const person1 = new Person("Hemendra", 23);

person1.greet();

person1.age = 25

console.log(person1.age)
```

```python
// Class Expression

const Person = class{
    constructor(name, age){
        this.name = name,
        this.age = age
    }
    display(){
        console.log(`Hello ${this.name}, You are of age ${this.age}`)
    }
}

person1 = new Person('Chandaluri', 25)
console.log(person1)
```

```python
// 3) Static Methods
// - Static methods belong to the class itself, not the instances.


class MathOperations{
    static add(a,b){
        return a + b
    }
}

console.log(MathOperations.add(2,4))
```

```python
// 4) Inheritance
// - Classes can inherit from other classes using the extends keyword.


class Animal{
    constructor(name){
        this.name = name 
    }

    speak(){
        console.log(`Animal named ${this.name} can speak and make noise`)
    }
}


class Dog extends Animal{
    speak(){
        console.log(`Dog named ${this.name} can bark`)
    }
}

puppy = new Dog('Rocky')
puppy.speak()
```

```python
// 5) Private Fields(ES2022)
// - Private fields are declared with a # prefix and are only accessible within the class.


class BankAccount{
    #balance = 0 

    deposit(amount){
        this.#balance += amount
        console.log(`Amount ${amount} deposited successfully`)
    }

    withdraw(amount){
        if (this.#balance >= amount){
            this.#balance -= amount
            console.log(`Amount ${amount} withdrawn successfully`)
            return
        }
        console.log("In-sufficient balance to withdraw")
    }

    get_balance(){
        return this.#balance
    }
}


const account1 = new BankAccount()
console.log(account1.get_balance())
account1.deposit(1000)
account1.withdraw(200)

console.log(account1.get_balance())
```

```python
// Freezing Objects

// 1) Shallow Freezing

const person = Object.freeze({
    name: "Hemendra",
    address:{
        city:'Hyderabad',
        pincode: 600084
    }
})


console.log(person)

person.name = 'Chandaluri'
person.address.city = 'Guntur'
console.log(person)





// 2) Deep Freezing


function deep_freeze(obj){
    Object.freeze(obj)

    Object.keys(obj).forEach((eachKey)=>{
        if(typeof obj[eachKey] === 'object' && obj[eachKey] !== null && !Object.isFrozen(obj[eachKey])){
            deep_freeze(obj[eachKey])
        }
    })

    return obj

}

const person = {
    name: "Hemendra",
    address:{
        city:'Hyderabad',
        pincode: 600084
    }
}

console.log(typeof person.name)
console.log(typeof person.address)
```
