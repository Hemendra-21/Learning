**1. `import`, `export`, `default` vs. named**

TypeScript uses ES6 module syntax for organizing code into reusable pieces. Here's how these concepts work:

----------


**Named Exports**

- You can export multiple items from a file using named exports.

- When importing, you must use the exact names of the exported items.

*Example:*

```python
export const add = (a: number, b: number): number => a + b;
export const subtract = (a: number, b: number): number => a - b;
```

```python
import { add, subtract } from './utils/mathUtils';

console.log(add(5, 3)); // Output: 8
console.log(subtract(5, 3)); // Output: 2
```

----

**Default Exports**

- A file can have only one default export.

- When importing, you can name the imported item anything you like.

*Example:*

```python
export default function log(message: string): void {
    console.log(message);
}
```

```python
import log from './utils/logger';

log('Hello, TypeScript!'); // Output: Hello, TypeScript!
```

---------

**Combining Named and Default Exports**

You can mix named and default exports in the same file.

*Example:*

```python
export const multiply = (a: number, b: number): number => a * b;
export default function divide(a: number, b: number): number {
    return a / b;
}
```

```python
import divide, { multiply } from './utils/mixedUtils';

console.log(multiply(6, 2)); // Output: 12
console.log(divide(6, 2));   // Output: 3
```

#### **2. Barrel Files (`index.ts` pattern)**

**What is a Barrel File?**

A barrel file is a single file (usually named `index.ts`) that re-exports items from multiple files in the same directory. This allows you to import everything from one central location instead of importing from individual files.




**Why Use Barrel Files?**

- **Simplifies Imports**: Instead of importing from multiple files, you can import everything from the barrel file.

- **Improves Code Organization**: Keeps your imports clean and manageable, especially in large projects.

- **Encapsulation**: Hides the internal structure of your modules.

**How Barrel Files Work**

*Example Without a Barrel File:*

Imagine you have the following folder structure:

```python
utils/
  mathUtils.ts
  stringUtils.ts
```

```python
// mathUtils.ts

export const add = (a: number, b: number): number => a + b;
export const subtract = (a: number, b: number): number => a - b;
```

```python
// stringUtils.ts

export const toUpperCase = (str: string): string => str.toUpperCase();
export const toLowerCase = (str: string): string => str.toLowerCase();
```

To use these utilities, you would need to import them individually:

```python
import { add, subtract } from './utils/mathUtils';
import { toUpperCase, toLowerCase } from './utils/stringUtils';

console.log(add(5, 3));           // Output: 8
console.log(toUpperCase('hello')); // Output: HELLO
```

*Example With a Barrel File:*

Now, create a `utils/index.ts` file (the barrel file):

**index.ts**

```python
export * from './mathUtils';
export * from './stringUtils';
```

This file re-exports everything from `mathUtils.ts` and `stringUtils.ts`. Now, you can import everything from the `utils` folder:

```python
import { add, subtract, toUpperCase, toLowerCase } from './utils';

console.log(add(5, 3));           // Output: 8
console.log(toUpperCase('hello')); // Output: HELLO
```

-------

**Key Points to Remember**

- *File Naming*: The barrel file is typically named `index.ts` because it is automatically resolved when you import from the folder.
    - For example, `import { add } from './utils';` automatically looks for `utils/index.ts`.

- *Re-Export Syntax*:

    - `export * from './file'`; re-exports everything from the file.
    - `export { specificExport } from './file';` re-exports specific items.
- *Avoid Circular Dependencies*: Be careful when using barrel files to avoid circular imports, which can cause runtime errors.
