##### Union Types (`number | string`)

Union types allow a variable to hold one of several types. It is useful when a value can be of multiple types.

Example:

```python
function printId(id: number | string): void {
    if (typeof id === "string") {
        console.log(`ID is a string: ${id.toUpperCase()}`);
    } else {
        console.log(`ID is a number: ${id}`);
    }
}

printId(101);        // Output: ID is a number: 101
printId("abc123");   // Output: ID is a string: ABC123
```

Use Case:

- **API Responses**: When an API can return multiple types (e.g., `string` or `number` for an ID).

- **Function Parameters**: When a function accepts multiple types of inputs.

##### Intersection Types (`A & B`)

Intersection types combine multiple types into one. A value must satisfy all the combined types.

Example:


```python
type Person = {
    name: string;
    age: number;
};

type Employee = {
    employeeId: number;
    department: string;
};

type Manager = Person & Employee;

const manager: Manager = {
    name: "Alice",
    age: 30,
    employeeId: 101,
    department: "HR",
};

console.log(manager);
```

Use Case:

- **Combining Models**: When an object needs to satisfy multiple interfaces or types (e.g., a Manager is both a Person and an Employee).

- **Extending Functionality**: Adding additional properties to existing types.

**Type Narrowing with Type Guards**

Type narrowing ensures that TypeScript understands the specific type of a variable at runtime. This is achieved using **type guards**.


```python
// Using `typeof`


function processValue(value: number | string): void {
    if (typeof value === "string") {
        console.log(`String value: ${value.toUpperCase()}`);
    } else {
        console.log(`Number value: ${value * 2}`);
    }
}

processValue(10);       // Output: Number value: 20
processValue("hello");  // Output: String value: HELLO
```

```python
// Using instanceof

class Dog {
    bark() {
        console.log("Woof!");
    }
}

class Cat {
    meow() {
        console.log("Meow!");
    }
}

function handleAnimal(animal: Dog | Cat): void {
    if (animal instanceof Dog) {
        animal.bark();
    } else {
        animal.meow();
    }
}

handleAnimal(new Dog()); // Output: Woof!
handleAnimal(new Cat()); // Output: Meow!
```

##### Using Custom Type Predicates

```python
type Fish = { swim: () => void };
type Bird = { fly: () => void };

function isFish(animal: Fish | Bird): animal is Fish {
    return (animal as Fish).swim !== undefined;
}

function handleAnimal(animal: Fish | Bird): void {
    if (isFish(animal)) {
        animal.swim();
    } else {
        animal.fly();
    }
}

const fish: Fish = { swim: () => console.log("Swimming...") };
const bird: Bird = { fly: () => console.log("Flying...") };

handleAnimal(fish); // Output: Swimming...
handleAnimal(bird); // Output: Flying...
```

Use Case:

- **Dynamic Behavior**: Handling different types in a single function.

- **Polymorphism**: Differentiating between classes or interfaces at runtime.
