**Class Syntax in TypeScript**

A class in TypeScript is defined using the `class` keyword. It can have properties, methods, and a constructor.



```python
class Person {
    name: string;
    age: number;

    constructor(name: string, age: number) {
        this.name = name;
        this.age = age;
    }

    greet(): void {
        console.log(`Hello, my name is ${this.name} and I am ${this.age} years old.`);
    }
}

const person = new Person("Alice", 30);
person.greet(); // Output: Hello, my name is Alice and I am 30 years old.
```

**Constructors and Member Visibility**

The `constructor` is a special method used to initialize class properties. Member visibility modifiers (`public`, `private`, `protected`) control access to class members.

- `public`: Accessible from anywhere (default).

- `private`: Accessible only within the class.
- `protected`: Accessible within the class and its subclasses.

```python
class Employee {
    public name: string;
    private salary: number;
    protected department: string;

    constructor(name: string, salary: number, department: string) {
        this.name = name;
        this.salary = salary;
        this.department = department;
    }

    public getSalary(): number {
        return this.salary; // Accessible within the class
    }
}

class Manager extends Employee {
    constructor(name: string, salary: number, department: string) {
        super(name, salary, department);
    }

    public getDepartment(): string {
        return this.department; // Accessible because it's protected
    }
}

const emp = new Employee("John", 50000, "IT");
console.log(emp.name); // Accessible
// console.log(emp.salary); // Error: Property 'salary' is private
```

**`readonly` Keyword**

The `readonly` modifier makes a property immutable after initialization (either in the declaration or in the constructor).

```python
class Car {
    readonly brand: string;

    constructor(brand: string) {
        this.brand = brand;
    }

    displayBrand(): void {
        console.log(`This car is a ${this.brand}.`);
    }
}

const car = new Car("Toyota");
car.displayBrand(); // Output: This car is a Toyota.
// car.brand = "Honda"; // Error: Cannot assign to 'brand' because it is a read-only property.
```

**`static` Members**

Static members belong to the class itself, not to any instance. They are accessed using the class name.



```python
class MathUtils {
    static PI: number = 3.14;

    static calculateCircumference(radius: number): number {
        return 2 * MathUtils.PI * radius;
    }
}

console.log(MathUtils.PI); // Output: 3.14
console.log(MathUtils.calculateCircumference(10)); // Output: 62.8
```

**Inheritance and `super()`**

- A class can extend another class using the `extends` keyword. 
- The `super()` function is used to call the parent class's constructor.


```python
class Animal {
    name: string;

    constructor(name: string) {
        this.name = name;
    }

    makeSound(): void {
        console.log("Some generic sound...");
    }
}

class Dog extends Animal {
    constructor(name: string) {
        super(name); // Call the parent class constructor
    }

    makeSound(): void {
        console.log(`${this.name} says: Woof!`);
    }
}

const dog = new Dog("Buddy");
dog.makeSound(); // Output: Buddy says: Woof!
```
