# Introduction to Transpilation and Compilation
Transpilation and compilation are essential processes in TypeScript development. Transpilation refers to converting TypeScript code into JavaScript, while compilation ensures that the TypeScript code adheres to type safety and other language rules. These processes enable developers to write modern, type-safe code while maintaining compatibility with JavaScript environments.

# Setting Up TypeScript Compiler
To start using TypeScript, you need to install the TypeScript compiler. This can be done using npm, the Node.js package manager.

Run the following command to install TypeScript globally:

```python
# Install TypeScript globally
!npm install -g typescript

# Verify the installation
!tsc --version
```

# Transpiling TypeScript to JavaScript
To transpile TypeScript code into JavaScript, create a simple TypeScript file and use the `tsc` command.

For example, create a file named `example.ts` with the following content:

```python
# example.ts
message: str = "Hello, TypeScript!"
print(message)
```

Then, run the following command to transpile the TypeScript file into JavaScript:

```python
# Transpile TypeScript to JavaScript
!tsc example.ts
```

# Configuring tsconfig.json
The `tsconfig.json` file allows you to customize the TypeScript compilation process. Create a `tsconfig.json` file in your project directory with the following content:

```python
# tsconfig.json
{
  "compilerOptions": {
    "target": "es6",
    "module": "commonjs",
    "strict": true,
    "outDir": "./dist"
  },
  "include": ["src/**/*"],
  "exclude": ["node_modules"]
}
```

# Watching Files for Changes
The TypeScript compiler provides a `--watch` flag to automatically transpile files whenever changes are detected. This is useful for development workflows.

Run the following command to enable watch mode:

```python
# Enable watch mode
!tsc --watch
```

# Handling Compilation Errors
TypeScript provides detailed error messages during compilation. Below are some common errors and how to resolve them:

1. **Type Mismatch**: Occurs when a variable is assigned a value of an incompatible type.
   - **Example**: Assigning a string to a number variable.
   - **Solution**: Ensure the types match or use type assertions.

2. **Missing Properties**: Occurs when an object is missing required properties.
   - **Solution**: Add the missing properties or update the type definition.

3. **Unknown Types**: Occurs when TypeScript cannot infer the type of a variable.
   - **Solution**: Explicitly define the type.

```python
# Example of a type mismatch error
let age: number = "twenty"; // Error: Type 'string' is not assignable to type 'number'.

# Corrected code
let age: number = 20;
```
