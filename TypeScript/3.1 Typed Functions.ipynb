##### 1. Type Parameters and Return Values

You can specify the types of function parameters and the return value explicitly. This is particularly useful for ensuring that the function behaves as expected.

Example:

```python
function add(a: number, b: number): number {
    return a + b;
}

const result = add(5, 10); // ✅ Valid
// const invalidResult = add(5, "10"); // ❌ Error: Argument of type 'string' is not assignable to parameter of type 'number'.
console.log(result); // Output: 15
```

**Best Case:**

- Use typed functions when working with mathematical operations, string manipulations, or any logic where parameter and return types are predictable.

-------

##### 2. Return Type Inference

TypeScript can infer the return type of a function based on its implementation. However, explicitly specifying the return type is a good practice for clarity and maintainability.

Example:



```python
function greet(name: string) {
    return `Hello, ${name}!`; // TypeScript infers the return type as `string`
}

const message = greet("Hemendra");
console.log(message); // Output: Hello, Hemendra!
```

```python
// Explicit Return Type:

function multiply(a: number, b: number): number {
    return a * b;
}

const product = multiply(4, 5); // Output: 20
```

**Best Case:**

- Use explicit return types for complex functions or when working in a team to make the code more understandable.

-------


##### 3. Combining Parameters and Return Types

You can combine multiple parameter types and return types using **union types** or **generics**.

Example with Union Types:

```python
function formatInput(input: string | number): string {
    return typeof input === "number" ? input.toFixed(2) : input.toUpperCase();
}

console.log(formatInput(42)); // Output: "42.00"
console.log(formatInput("typescript")); // Output: "TYPESCRIPT"
```

**Best Case:**

- Use union types when a function can accept multiple types of inputs but always returns a consistent type.

-------

##### 4. Advanced Example with Generics

Generics allow you to create reusable, type-safe functions that work with multiple types.

Example:



```python
function identity<T>(value: T): T {
    return value;
}

const num = identity<number>(42); // T is inferred as number
const str = identity<string>("TypeScript"); // T is inferred as string

console.log(num); // Output: 42
console.log(str); // Output: TypeScript
```

**Best Case:**

- Use generics for utility functions like `identity`, `map`, or `filter` where the type depends on the input.
