Generics in TypeScript allow you to create reusable and flexible components, such as functions, classes, or interfaces, that can work with a variety of types instead of being restricted to one specific type. They provide a way to write code that is both type-safe and adaptable.

----------

**1. What Are Generics?**

Generics are placeholders for types. Instead of hardcoding a specific type, you use a "type parameter" (commonly represented as `<T>`). This allows the type to be determined when the code is used, making it reusable for different types.

*For example:*



```python
function identity<T>(value: T): T {
    return value;
}

// Usage:
const num = identity<number>(42); // T is number
const str = identity<string>("Hello"); // T is string
```

Here, T is a generic type parameter. When calling the function, you specify the type (number, string, etc.), and TypeScript ensures type safety.

--------

**2. Why Use Generics?** 

- **Reusability**: Write code once and use it for multiple types.
- **Type Safety**: Catch type-related errors at compile time.
- **Flexibility**: Handle a variety of types without duplicating code.
---------

**3. Generic Functions**

A generic function works with any type. The type is specified when the function is called.

```python
function wrapInArray<T>(value: T): T[] {
    return [value];
}

// Usage:
const numArray = wrapInArray<number>(5); // [5]
const strArray = wrapInArray<string>("TypeScript"); // ["TypeScript"]
```

--------

**4. Constraining Generics**

Sometimes, you want to restrict the types that a generic can accept. You can use a constraint with the `extends` keyword.

```python
function getLength<T extends { length: number }>(item: T): number {
    return item.length;
}

// Usage:
const strLength = getLength("Hello"); // Works, string has a length property
const arrLength = getLength([1, 2, 3]); // Works, array has a length property
// const numLength = getLength(42); // Error: number doesn't have a length property
```

Here, `T` is constrained to types that have a `length` property.

----------

**5. Generics with Interfaces**

You can use generics in interfaces to define reusable type-safe structures.



```python
interface Box<T> {
    content: T;
}

const stringBox: Box<string> = { content: "Hello" };
const numberBox: Box<number> = { content: 42 };
```

------

**6. Generics with Classes**

Generics can also be used in classes to create flexible and reusable data structures.

```python
class DataStore<T> {
    private data: T[] = [];

    add(item: T): void {
        this.data.push(item);
    }

    getAll(): T[] {
        return this.data;
    }
}

// Usage:
const stringStore = new DataStore<string>();
stringStore.add("TypeScript");
console.log(stringStore.getAll()); // ["TypeScript"]

const numberStore = new DataStore<number>();
numberStore.add(42);
console.log(numberStore.getAll()); // [42]
```

-----

**7. Key Points to Remember**

- `<T>` is just a convention; you can use any name (e.g., `<U>`, `<Type>`).
- Generics can have multiple type parameters: `<T, U>`.
- They make your code more reusable and type-safe.
------------

**8. Real-World Example**

Imagine a function to fetch data from an API. Generics ensure the response is strongly typed:



```python
async function fetchData<T>(url: string): Promise<T> {
    const response = await fetch(url);
    return response.json();
}

// Usage:
interface User {
    id: number;
    name: string;
}

const user = await fetchData<User>("https://api.example.com/user/1");
console.log(user.name); // TypeScript knows `user` has a `name` property
```
