Arrays and Tuples in TypeScript


1) Arrays

    In TypeScript, arrays are used to store multiple values of the same type. You can define arrays in two ways:


```python
// 1) Using square brackets ([]):

let numbers: number[] = [1, 2, 3, 4];
let strings: string[] = ["apple", "banana", "cherry"];


// 2) Using the Array<Type> syntax:

let numbers: Array<number> = [1, 2, 3, 4];
let strings: Array<string> = ["apple", "banana", "cherry"];
```

Key Features:

- Arrays are strongly typed, meaning you can only store values of the specified type.

- TypeScript will throw an error if you try to add a value of a different type.

Example:



```python
let fruits: string[] = ["apple", "banana"];
fruits.push("cherry"); // ✅ Allowed
// fruits.push(42); // ❌ Error: Argument of type 'number' is not assignable to type 'string'.
```

------------

2. Tuples

    Tuples are a special type of array in TypeScript that allow you to define an array with a fixed number of elements, where each element can have a different type.

Syntax:

```python
let tuple: [Type1, Type2, Type3];

// ex:
let person: [string, number, boolean] = ["Alice", 30, true];
```

Key Features:

- Tuples have a fixed length and order.
- Each element in the tuple must match the specified type.

Accessing Tuple Elements: You can access tuple elements using their index:



```python
console.log(person[0]); // Output: "Alice"
console.log(person[1]); // Output: 30
```

Optional Elements in Tuples: You can make some elements optional:



```python
let optionalTuple: [string, number?] = ["Alice"];
```

Rest Elements in Tuples: 

You can use rest elements to allow a variable number of elements of a specific type:



```python
let restTuple: [string, ...number[]] = ["Alice", 1, 2, 3];
```
