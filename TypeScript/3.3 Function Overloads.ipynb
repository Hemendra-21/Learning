##### **1. Declaring Multiple Signatures**

You can declare multiple function signatures (overloads) for a single function. These signatures describe the possible ways the function can be called.

Example:



```python
// Overload signatures
function getInfo(id: number): string;
function getInfo(name: string): string;

// Implementation
function getInfo(param: number | string): string {
    if (typeof param === "number") {
        return `User ID: ${param}`;
    } else {
        return `User Name: ${param}`;
    }
}

console.log(getInfo(123));       // Output: User ID: 123
console.log(getInfo("Hemendra")); // Output: User Name: Hemendra
// console.log(getInfo(true));    // ❌ Error: No overload matches this call
```

Key Points:

- The **overload signatures** define the valid ways to call the function.

- The **implementation** must handle all possible cases defined in the overload signatures.

##### **2. Implementation with Conditional Logic**

The implementation of the function must use type guards (e.g., `typeof`, `instanceof`) or other logic to differentiate between the input types and handle them accordingly.

Example with Arrays:



```python
// Overload signatures
function combine(input1: number[], input2: number[]): number[];
function combine(input1: string[], input2: string[]): string[];

// Implementation
function combine(input1: number[] | string[], input2: number[] | string[]): number[] | string[] {
    if (typeof input1[0] === "number" && typeof input2[0] === "number") {
        return [...input1, ...input2];
    } else if (typeof input1[0] === "string" && typeof input2[0] === "string") {
        return [...input1, ...input2];
    }
    throw new Error("Invalid input types");
}

console.log(combine([1, 2], [3, 4]));       // Output: [1, 2, 3, 4]
console.log(combine(["a", "b"], ["c", "d"])); // Output: ["a", "b", "c", "d"]
// console.log(combine([1, 2], ["a", "b"]));  // ❌ Error: Invalid input types
```

##### Best Practices

**1. Prefer Function Expressions with Types**

Function expressions allow you to define a function and assign it to a variable. Adding explicit types ensures type safety and better readability.

*Example:*


```python
// Function expression with explicit types
const add: (a: number, b: number) => number = function (a, b) {
    return a + b;
};

console.log(add(2, 3)); // Output: 5
// console.log(add(2, "3")); // ❌ Error: Argument of type 'string' is not assignable to parameter of type 'number'.
```

**2. Use Arrow Functions for Callbacks**

Arrow functions are concise and automatically bind this to the surrounding context, making them ideal for callbacks.

*Example:*

```python
// Define an array of numbers
const numbers: number[] = [1, 2, 3, 4, 5];

// Use map with an arrow function and inferred types
const squaredNumbers: number[] = numbers.map((num: number): number => num * num);

console.log(squaredNumbers); // Output: [1, 4, 9, 16, 25]
```

**3. Use void for Event Handlers**

Event handlers typically do not return a value. Explicitly using void as the return type makes this clear and avoids unintended behavior.

*Example:*


```python
// Event handler with `void` return type
const handleClick: (event: MouseEvent) => void = (event) => {
    console.log("Button clicked!", event);
};

// Simulating a button click
document.querySelector("button")?.addEventListener("click", handleClick);
```
