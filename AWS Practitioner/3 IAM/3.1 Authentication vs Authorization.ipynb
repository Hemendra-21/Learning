### **🔐 Authentication vs Authorization (Clear Difference)**

| Aspect              | Authentication            | Authorization                   |
| ------------------- | ------------------------- | ------------------------------- |
| Question it answers | **Who are you?**          | **What are you allowed to do?** |
| Happens first?      | ✅ YES                     | ⏭️ After authentication         |
| Focus               | Identity verification     | Permission checking             |
| Example             | Login with password + OTP | Access S3, EC2, DynamoDB        |
| AWS service         | IAM, Cognito              | IAM Policies                    |


---------

**🔑 Authentication (WHO are you?)**

**📖 Definition**

Authentication is the process of **verifying identity.**

AWS checks:

> “Are you really Hemendra?”


------------

**🔍 How Authentication Works in AWS**

AWS authenticates you using:

- Username + Password

- Access Key + Secret Key

- MFA (OTP / hardware key)

- IAM Role (temporary credentials)

📌 If authentication fails → **access denied immediately**

------------------


**🧠 AWS Examples**

**Example 1: Console Login**

You enter:

- IAM username

- Password

- MFA code

👉 AWS verifies identity

**👉 Authentication successful**


-------------

**Example 2: AWS CLI**


```python
aws s3 ls
```

AWS checks:

- Access Key

- Secret Key

If keys are valid → authentication success

-------------

**🚫 Authentication Does NOT Decide Permissions**

Even after successful authentication:

- You **may or may not** have permission to act

----

#### **🛂 Authorization (WHAT are you allowed to do?)**

**📖 Definition**

Authorization determines **which actions you can perform on which resources.**

AWS asks:

> “You are Hemendra, but what can you do?”

-------------

**🔍 How Authorization Works in AWS**

AWS evaluates:

- IAM Policies

- Effect (Allow / Deny)

- Action

- Resource

- Conditions (if any)

📌 Authorization happens **AFTER authentication**


-----------

**🧠 AWS Example**

You run:

```python
aws ec2 start-instances --instance-id i-123
```

AWS checks:

1) Are you authenticated? ✅

2) Do your policies allow `ec2:StartInstances`?

    - If YES → action allowed

    - If NO → **AccessDenied**

------------

**❌ Authorization Failure Example**

You are logged in successfully, but:

```python
"Action": "ec2:DescribeInstances"
```

You try to **start EC2 ❌**

➡️ Authorization fails

---------

**🔁 Authentication vs Authorization (Flow in AWS)**

**Step-by-step Flow**

**1) Authenticate user**

**2) Evaluate IAM policies**

**3) Check explicit denies**

**4) Allow or deny action**

📌 No authentication → no authorization

📌 Authentication alone ≠ access

-------------

**🌍 Real-World Analogy (Very Easy to Remember)**

**🏢 Office Building**

- **Authentication** → ID card check at entrance

- **Authorization** → Which rooms you can enter

You may enter the building, but:

- ❌ Server room

- ❌ CEO cabin

-------------

**☁️ AWS-Specific Mapping**

| AWS Component               | Role           |
| --------------------------- | -------------- |
| IAM User / Role             | Authentication |
| Password / Access Key / MFA | Authentication |
| IAM Policy                  | Authorization  |
| Effect, Action, Resource    | Authorization  |
| Explicit Deny               | Authorization  |
