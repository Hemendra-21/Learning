### **🔐 AWS IAM (Identity and Access Management)**

**What is IAM?**

**AWS Identity and Access Management (IAM)** is an AWS service that helps you **control who can access AWS resources and what actions they can perform.**

In simple words:

> **IAM = Authentication (Who you are) + Authorization (What you can do)**

It is the **security backbone** of AWS.

---------------


**Why IAM Matters (Real-World Context)**

Imagine AWS as a **company building:**

- **Users** → Employees

- **Groups** → Departments (Dev, QA, Admin)

- **Roles** → Temporary job responsibilities

- **Policies** → Permission rules

- **MFA** → Biometric / OTP security

Without IAM, **anyone could delete servers or access sensitive data 😬**


---------------




### **Core Components of IAM**

**1️⃣ IAM Users**

An **IAM User** represents a **person or application.**

Each user can have:

- **Console access** (username + password)

- **Programmatic access** (Access Key + Secret Key)

📌 Example:

- You → IAM User

- Jenkins CI → IAM User (programmatic)

----------------

**2️⃣ IAM Groups**

A **Group** is a collection of users with **same permissions.**

📌 Example:

- `Developers` group → EC2 + S3 access

- `Admins` group → Full AWS access

👉 Permissions are **assigned to groups**, not users directly (best practice).

-------------------


**3️⃣ IAM Policies (MOST IMPORTANT)**

Policies define **permissions.**

They are written in **JSON** and answer:

- **What action?**

- **On which resource?**

- **Allowed or denied?**

📌 Example:

```python
{
  "Effect": "Allow",
  "Action": "s3:PutObject",
  "Resource": "arn:aws:s3:::my-bucket/*"
}
```

**🧠 Types of policies:**

- **AWS Managed** (provided by AWS)

- **Customer Managed** (you create)

- **Inline** (attached to a single user/role)

----------------

**4️⃣ IAM Roles**

A **Role** is used when **one AWS service needs to access another.**

No username or password ❌

Uses **temporary credentials ✅**

📌 Examples:

- EC2 → access S3

- Lambda → access DynamoDB

- Cross-account access

👉 Very important for **secure architectures.**

--------------

**5️⃣ Authentication vs Authorization**

| Concept        | Meaning                     |
| -------------- | --------------------------- |
| Authentication | Who are you?                |
| Authorization  | What are you allowed to do? |

IAM handles **both.**

------------

**6️⃣ MFA (Multi-Factor Authentication)**

Extra security layer:

- Password + OTP (Google Authenticator, Authy, etc.)

📌 Best practice:

- **Enable MFA for root user**

- Enable MFA for **admins**

----------------


**AWS Root User (⚠️ Special Case)**

When you create an AWS account:

- Root user has **FULL ACCESS**

- Should be used **only for account setup**

🚫 Never use root user for daily work

✅ Create IAM users instead

-------------------

**IAM Key Characteristics** 

✔ Global service (not region-specific)

✔ Free of cost

✔ Fine-grained permissions

✔ Supports **least privilege principle**

✔ Centralized access control

------------

**High-Level Flow (How IAM Works)**

- User/service makes a request

- IAM checks **identity**

- IAM evaluates **policies**

- Allow or deny access

---------------

**Best Practices (VERY IMPORTANT)**

✅ Use **groups**, not individual permissions

✅ Follow **least privilege**

✅ Enable **MFA**

✅ Rotate access keys

✅ Use **roles** instead of hard-coding keys

✅ Never share credentials


-----

**How This Connects to Other AWS Services**

- EC2 → IAM Role

- S3 → IAM Policy

- Lambda → IAM Role

- RDS → IAM Authentication

- CloudTrail → IAM auditing
