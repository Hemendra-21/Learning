### **🔐 IAM Policy Basics (Core of AWS Security)**

IAM Policies define **WHAT actions are allowed or denied on WHICH AWS resources and UNDER WHAT CONDITIONS.**

Think of a policy as a **permission rulebook** attached to:

- a **User**

- a **Group**

- a **Role**

-------------



#### **🧾 IAM Policy JSON Structure**

IAM policies are written in **JSON.**

**Basic Skeleton**

```python
{
  "Version": "2012-10-17",
  "Statement": [
    {
      "Effect": "",
      "Action": "",
      "Resource": ""
    }
  ]
}
```

Let’s break each part.

-----------

**1️⃣ Version**

```python
"Version": "2012-10-17"
```

- This is **NOT the policy version**

- It indicates the **policy language version**

- Always use `2012-10-17` (AWS best practice)

--------------

**2️⃣ Statement (Heart of the Policy)**

A policy can have **one or multiple statements.**

```python
"Statement": [
  { ... },
  { ... }
]
```

Each statement is evaluated independently.

--------

**🎯 Effect, Action, Resource (Most Important)**

**🔹 Effect**

Defines whether you are **allowing or denying** an action.

```python
"Effect": "Allow"
```

or

```python
"Effect": "Deny"
```

**📌 Explicit DENY always wins**, even if another policy allows it.

-------------

**🔹 Action**

Specifies **what AWS actions are allowed/denied.**

```python
"Action": "s3:ListBucket"
```

Multiple actions:

```python
"Action": [
  "s3:GetObject",
  "s3:PutObject"
]
```

Wildcards:

```python
"Action": "s3:*"
```

➡️ Allows **ALL S3 actions** (dangerous if misused)

------------

**🔹 Resource**

Specifies **which AWS resource** the action applies to.

```python
"Resource": "arn:aws:s3:::my-bucket-name"
```

For **objects inside a bucket:**

```python
"Resource": "arn:aws:s3:::my-bucket-name/*"
```

Wildcard (ALL resources):

```python
"Resource": "*"
```

**🚨 Avoid unless absolutely required**

-----------

#### **✅ Complete Example: Read-only S3 Access**


```python
{
  "Version": "2012-10-17",
  "Statement": [
    {
      "Effect": "Allow",
      "Action": [
        "s3:ListBucket",
        "s3:GetObject"
      ],
      "Resource": [
        "arn:aws:s3:::my-bucket",
        "arn:aws:s3:::my-bucket/*"
      ]
    }
  ]
}
```

👉 User can **view bucket & download files**

👉 Cannot upload or delete

----------------

#### **📦 Managed Policies vs Inline Policies**

**🟦 Managed Policies**

Policies **created once and reused.**

**Types:**

**1) AWS Managed** – created by AWS

Example:

- `AmazonS3ReadOnlyAccess`

- `AdministratorAccess`

**2) Customer Managed** – created by YOU

**Example:**

Attach same policy to:

- Dev user

- QA user

- EC2 role

✅ Easy to manage

✅ Reusable

✅ Recommended for production

--------------

**🟥 Inline Policies**

Policies **embedded directly** into a single user, group, or role.

📌 Characteristics:

- One-to-one relationship

- Cannot be reused

- Deleted automatically when entity is deleted

🚫 Not recommended for large environments

--------------



**🔍 When to Use What?**

| Scenario                     | Use              |
| ---------------------------- | ---------------- |
| Company-wide access rules    | Managed Policy   |
| Temporary one-off permission | Inline Policy    |
| Production environment       | Managed Policy ✅ |


--------------


#### **🛡️ Principle of Least Privilege (VERY IMPORTANT)**

**📖 Definition**

> Give **only the minimum permissions required** to perform a task—nothing more.

-------


**❌ Bad Practice**

```python
"Action": "*",
"Resource": "*"
```

➡️ Full admin access (security risk)

----------

**✅ Good Practice**

```python
"Action": "ec2:DescribeInstances",
"Resource": "*"
```

➡️ Only view EC2 instances

-----------

**🧠 Real-World Example**

Role: **Data Analyst**

- Needs to read data from S3

- DOES NOT need EC2, IAM, VPC access

So permissions should be:

- ✅ s3:GetObject

- ❌ ec2:*

- ❌ iam:*

------------

#### **🏆 IAM Best Practices (Exam + Real World)**

----------

**🔐 1️⃣ Enable MFA (Multi-Factor Authentication)**

MFA = **Password + Extra verification**

Examples:

- Authenticator app

- Hardware MFA device

📌 Always enable MFA for:

- Root user

- Admin users

Even if password leaks → attacker **cannot log in.**

----------

**🚫 2️⃣ Avoid Using Root User**

Root user:

- Has **FULL access**

- Cannot be restricted by **IAM policies**

**📌 Best practice:**

- Use root user ONLY for:

    - Account setup

    - Billing

    - Enable MFA

- Create **Admin IAM** User for daily work


-----------

**🔄 3️⃣ Rotate Credentials Regularly**

Access keys should be rotated:

- Every 90 days (recommended)

Why?

- Reduces damage if keys are leaked

- Limits long-term misuse

**📌 Best approach:**

- Use **IAM Roles** instead of access keys wherever possible (EC2, Lambda)
