### **📌 Day 1 — Refresh Foundations (Quick Recap)**

- **Goals**:

    - Set up Python environment (`venv`, `pip`, `requirements.txt`)

    - Revisit API calls using `requests` (example: call OpenAI or Hugging Face API)

    - Quick Git refresher (clone, branch, commit, push, PR)

- **Task**:

    - Write a small Python script that calls an API (e.g., OpenAI or weather API) and prints the response.

    - Push this script to GitHub.

------

### **📌 Days 2–4 — Understanding LLMs (Deep Dive)**

- **Goals:**

    - Learn how LLMs work (tokens, embeddings, context window, temperature, top-p).

    - Learn about different models (OpenAI GPT, Anthropic Claude, Meta LLaMA, Mistral).

    - Understand prompting techniques: zero-shot, few-shot, chain-of-thought, role prompting.

- **Tasks:**

    - Try OpenAI Playground (or Hugging Face Inference API).

    - Experiment with prompts: ask same question with different styles.

    - Compare GPT vs Hugging Face models.

- **Mini Project:**

    - Create a Python script that summarizes an article (input: text, output: summary via API).

-----------

### **📌 Days 5–8 — LangChain Fundamentals**

- **Goals:**

    - Install LangChain and learn its core building blocks:

        - 1) **LLMs** → connecting to OpenAI/Hugging Face.

        - 2) **Prompt Templates** → structured prompts with variables.

        - 3) **Chains** → combining multiple steps.

        - 4) **Memory** → making chatbots remember past context.

    - Understand difference between **LLMChain** vs **ConversationChain**.

- **Tasks:**

    - Build:

        - A simple Q&A bot (LangChain + OpenAI).

        - A chatbot with memory (LangChain ConversationChain).

- **Mini Project:**

    - News summarizer bot (fetch latest news → summarize with LangChain).

-------------

### **📌 Days 9–12 — Advanced LangChain (Tools + RAG)**

- **Goals:**

    - Learn **Tools & Agents** in LangChain:

        - Agents = LLMs that decide which tools to call.

        - Tools = search, calculator, APIs, code execution.

    - Learn **Vector Databases** (FAISS, Pinecone, Chroma).

    - Learn **RAG (Retrieval Augmented Generation)** → fetch documents + answer queries.

- **Tasks:**

    - Implement a document Q&A bot (upload PDF → ask questions → get answers).

    - Use FAISS for vector storage and embeddings.

- **Mini Project:**

    - PDF-based chatbot (upload resume → chatbot answers HR questions about you).

----------

### **📌 Days 13–15 — LangGraph (Orchestration & Workflows)**

- **Goals:**

    - Understand LangGraph vs LangChain.

        - LangChain = components.

        - LangGraph = orchestrator for building workflows like state machines.

    - Learn about **Nodes, Edges, State, Checkpoints.**

    - Build async, multi-step LLM workflows.

- **Tasks:**

    - Build a workflow with multiple steps (e.g., “Research → Summarize → Format as Report”).

    - Add error handling + retries with LangGraph.

- **Mini Project:**

    - Research assistant: input a topic → LangGraph fetches articles → summarizes → outputs a structured report.

---------

### **📌 Days 16–17 — Combining Everything**

- **Goals:**

    - Connect LangChain + LangGraph.

    - Implement **multi-agent systems** (e.g., one bot for research, another for coding, another for summarizing).

- **Tasks:**

    - Multi-step pipeline:

        - User asks a question → LLM searches → retrieves docs → summarizes → formats.

- **Mini Project:**

    - A "Research + Report" pipeline where one agent gathers data, one verifies facts, one writes output.

-----------

### **📌 Days 18–20 — Open-Source Project Contribution**

- **Goals:**

    - Pick an open-source LangChain/LangGraph project from GitHub.

    - Understand repo structure.

    - Add new feature or modify existing one.

- **Steps:**

    - 1) Clone repo & run locally.

    - 2) Read README + docs to understand functionality.

    - 3) Identify a simple feature to add (e.g., new tool, new agent, improve UI, add dataset).

    - 4) Implement & test.

    - 5) Push changes to GitHub.

- **Suggested Open-Source Projects:**

    - LangChainHub - source not found

    - *LangGraph Examples* : https://github.com/langchain-ai/langgraph?utm_source=chatgpt.com

    - *Haystack (alternative RAG framework)* : https://github.com/deepset-ai/haystack
